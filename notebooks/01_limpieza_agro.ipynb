```python
import os

nombre_buscado = "Estimacion_Agricola Crudo.xlsx"
ruta_encontrada = None

# Buscar el archivo en Google Drive
for raiz, directorios, archivos in os.walk('/content/drive/MyDrive'):
    if nombre_buscado.lower() in [f.lower() for f in archivos]:
        for archivo in archivos:
            if archivo.lower() == nombre_buscado.lower():
                ruta_encontrada = os.path.join(raiz, archivo)
                break

if ruta_encontrada:
    print(f" ARCHIVO ENCONTRADO EN:\n{ruta_encontrada}")
else:
    print(" No se encontró el archivo con ese nombre exacto. Revisa el paso 3.")
```

```python
import pandas as pd

# Ruta exacta donde se encontró el archivo
ruta = '/content/drive/MyDrive/Proyecto Agro (Grupo 1)/Colabs/data/Estimacion_Agricola Crudo.xlsx'

# Cargar el archivo Excel
df_produccion = pd.read_excel(ruta)

# Visualizar las primeras filas
df_produccion.head()
```

### Verificamos los valores nulos y toma de decision

Encontramos que habia 8 valores nulos en las columnas `provincia`, `provincia_id`, `departamento` y `departamento_id`, entonces verificamos esas filas y vemos que decidimos hacer con ellas

```python
df_produccion[df_produccion['provincia'].isnull()]
```

Las 8 filas sin provincia corresponden a registros incompletos de 1979-1980

Al ser registros incompletos y representar una pequeña porción del dataset, la decisión más adecuada es eliminarlos.

```python
df_produccion.dropna(subset=['provincia', 'provincia_id', 'departamento', 'departamento_id'], inplace=True)
print('Filas con valores nulos eliminadas. Ahora verificamos si hay nulos restantes:')
display(df_produccion.isnull().sum())
```

```python
df_produccion.shape
```

### Convertir los IDs de float a entero

En el analisis exploratorio vimos que la columna `departamento_id` y `provincia_id` los datos estaban en tipo float y los debemos convertir a entertos

```python
df_produccion['provincia_id'] = df_produccion['provincia_id'].astype(int)
df_produccion['departamento_id'] = df_produccion['departamento_id'].astype(int)

df_produccion.dtypes
```

`provincia_id` y `departamento_id` convertidos de `float64` a `int64` con `.astype(int)`.
Ya no muestran el `.0` decimal que no tenía sentido para un código identificador.
