```python
import os

nombre_buscado = "clima_nasa_power_anual_por_provincia.xlsx"
ruta_encontrada = None

# Buscar el archivo en Google Drive
for raiz, directorios, archivos in os.walk('/content/drive/MyDrive'):
    if nombre_buscado.lower() in [f.lower() for f in archivos]:
        for archivo in archivos:
            if archivo.lower() == nombre_buscado.lower():
                ruta_encontrada = os.path.join(raiz, archivo)
                break

if ruta_encontrada:
    print(f" ARCHIVO ENCONTRADO EN:\n{ruta_encontrada}")
else:
    print(" No se encontró el archivo con ese nombre exacto. Revisa el paso 3.")
```

```python
import pandas as pd

# Ruta exacta donde se encontró el archivo
ruta = '/content/drive/MyDrive/Proyecto Agro (Grupo 1)/Colabs/data/clima_nasa_power_anual_por_provincia.xlsx'

# Cargar el archivo Excel
df_clima = pd.read_excel(ruta)

# Visualizar las primeras filas
df_clima.head()
```

## Verificamos Valores nulos y cambio de valor

```python
df_clima[df_clima['ALLSKY_SFC_SW_DWN'].isnull()]
```

```python
df_clima['ALLSKY_SFC_SW_DWN'] = df_clima['ALLSKY_SFC_SW_DWN'].fillna(0)
display(df_clima.head())
```

Verificamos que los valores nulos fueron reemplazados:

```python
df_clima.isnull().sum()
```

## Eliminacion de la provincia `Ciudad Autonoma de Buenos Aires`

```python
df_clima = df_clima[df_clima['provincia'] != 'Ciudad Autónoma de Buenos Aires']
display(df_clima.head())
print(f"Nueva forma del DataFrame después de la eliminación: {df_clima.shape}")
```

```python
df_clima['provincia'].unique()
```

Eliminamos la provincia `Ciudad Autónoma de Buenos Aires` y pasamos de tener 1080 filas a 1035

### Cambio de nombre de la columna `anio` por `año` como hicimos en el dataset 'Agro'

```python
df_clima.rename(columns={'anio': 'año'}, inplace=True)
display(df_clima.head())
```
